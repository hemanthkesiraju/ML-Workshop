```python
import plotly.graph_objects as go
# Let's visualize the histogram
trace = go.Histogram(
    x=data['Age'],
    nbinsx=20,
    marker={'color': 'red'})

# Layout
layout = go.Layout(
    title="Age Distribution"
)

# Combine the trace with the layout
fig = go.Figure(data=[trace], layout=layout)

# Display the figure
fig.show()
```
